## Problem 2 — BMI with keyword-only flag
Write `bmi(weight_kg, height_m, *, rounded=False)` and return BMI value, optionally rounded to 1 decimal place when `rounded=True`.
Test it with a couple examples, including weight_kg = 70, height_kg = 1.82 (once rounded and once without rounding).

Recall the formula for bmi: weight_kg/(height_m)^2.

```python
# TODO: Define bmi(...) and test with a couple examples, including weight_kg = 70, height_kg = 1.82. Do this once with rounding and once without rounding

#description for user comprehension
print("this is a bmi calcluator! It is designed to estimate body fat in relation to weight and height")

#setting weight and height as floats determined by user input
weight_kg = float(input("please select a value for weight in Kg"))
height_m = float(input("please select a value for height in m"))

#printing the inputted weight and height for user comprehension
print("your height is", height_m, "and your weight is", weight_kg,)

#creating a function accepting values for weight and height (inputted earlier)
#return value gives the function an operation to compute using the accepted values
def bmi(weight_kg, height_m):
    return (weight_kg/((height_m)**2))

#set the default rounding setting as a boolean false
rounded= False

#give user the option to input a string, letting them choose a rounding setting
rounding_select=str(input("would you like the Bmi value rounded to one decimal place? Input yes or no"))

#if user prints the string "yes", the boolean will change to true
if rounding_select=="yes":
    rounded=True

#if user prints the string "no" the boolean will remain false
if rounding_select=="no":
    rounded=False

#if user selects yes, and rouded becomes true, the final value will be rounded
if rounded == True:
    print("your bmi is",(float(round(bmi(weight_kg,height_m),1))))
    break

#if rounded is false, the final value will not be rounded
else rounded == False:
    print("your bmi is", (float(bmi(weight_kg,height_m))))


```

## Problem 4 — Quadratic roots
Implement `quad_roots(a, b, c)` returning the tuple of real roots of $ax^2+bx+c=0$ (handle 0, 1, or 2 real roots).  
(You may skip complex roots or return `None` if discriminant < 0.)

Test your function on the following polynomials:
* $3x^2-12x+12$
* $x^2-2x-3$
* $x^2 -2x + 3$

```python
# TODO: Implement quad_roots(a, b, c) for real roots. Use math.sqrt for discriminant >= 0.

#description for user readability
print("please input values for a, b, c,")

#letting user input values for a b and c
a=int(input("please input a value for the leading coefficient"))
b=int(input("please input a value for the second coefficient"))
c=int(input("please input a value for the constant"))

#displaying a function based on input
print("your function is", a,"x^2","+", b, "x","+",c,)

#creating a function which will take inputs for a b and c
def quad_roots(a, b, c):
    #defining the discriminant as the value under the root in quadratic formula
    #this makes it recyclable in code and easy to input
    disc = b*b - 4*a*c

    #if loop for if discriminant is under zero
    #returns no value (since there are no real roots)
    if disc < 0:
        return None
    
    #creating a value for the root of the discriminant (discriminant defined earlier to the power 1/2)
    #return statement: returns values for both quadratic formula possibilities
    disc_root = disc**0.5
    return ((-b - disc_root/(2*a), (-b + disc_root)/(2*a)))

#displays roots for user readablility
print("your root(s) are", quad_roots(a, b, c,))

```
