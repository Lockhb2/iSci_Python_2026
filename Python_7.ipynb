## Problem 1 — Read and summarize
Open `sample_numbers.txt`, read integers line-by-line, and compute:
- count, min, max, mean

Print the results.

```python

# TODO: Implement the file read & summary.
# Reminders (see Mini-exercise discussed in Files and Modules video):
# with open("sample_numbers.txt", "r", encoding="utf-8") as f:
#     for line in f:
#         ...
#with open ensures the file is closed after code is executed,


from os import path, read 

#setting nums equal to the values within this set
nums=[]

#with open will open file and close file when all code has run
with open("sample_numbers.txt", "r", encoding="utf-8") as f:
     for line in f:

           #removes whitespace from file
           s = line.strip()

           # if there is no line, continue anyways
           #prevents code from stopping
           if not s:
            continue

           # converting the data points form strings into 
           nums.append(float(s))


     #finding the various values required 
     print("the numbers in this file are:", nums)
     print("this many numbers are present:", len(nums))
     print("the minimum of these numbers is", min(nums))
     print("the maximum of these numbers is", max(nums))
     print("the mean of these numbers is", sum(nums)/len(nums))
  

```
