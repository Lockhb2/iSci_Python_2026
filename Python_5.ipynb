## Problem 1 — List toolbox
Start with `nums = [3, 1, 2]`. Do the following with **list methods** and print after each step:
- Append 5
- Extend with `[2, 2,2]`
- Remove a single `2`
- Count how many `2`s remain (and print this).
- Sort the list ascending
- Sort the list descending
- Recast the list as a tuple.

```python
# TODO: Use list methods to transform nums as described.
nums = [3, 1, 2]
# Your code here...

#.append will add a number to the end of this list
nums.append (5)
print("the number 5 has been appended ->", nums)

#.extend will add a set of numbers to the end of this list
nums.extend([2, 2, 2,])
print("the numbers [2,2,2,] has been added ->", nums)

#.remove will remove the first 2 in the list
nums.remove (2)
print("the first 2 has been removed ->", nums)

#.count will return the amount of a certain value in a list
print(" the number of 2s in your list is now ->", nums.count(2))

#.sort orders the list in ascending order
nums.sort()
print("your list has been sorted in ascending order ->", nums)

#.reverse will reverse the list. This sorts the list in descending order, since it was previously in ascending order.
nums.reverse()
print("your list in descending order is ->", nums)

#the tuple function converts the list to a tuple
print("the final tuple is", tuple(nums))


```

## Problem 2 — Set operations
Given `A = {1, 2, 3, 3, 2}` and `B = {3, 4, 5}`, compute:
- Union
- Intersection
- Difference `A - B`
- Symmetric difference
- Recast `A` as a list and as a tuple.

Print the result of each computation.

```python

# TODO: Perform set operations and print results.
A = {1, 2, 3, 3, 2}
B = {3, 4, 5}
# Your code here...

#printing A and B for user comprehension
print("the list A is", A, "and the list B is", B)

# "|" will return a list of all numbers in A and B, ignoring duplicates
print("the union of A and B is", A|B)

# "&" will return all numbers which A and B have in common
print("the intersection of A and B is", A & B)

# "-" will return all numbers whic are contained in A but excluded from B
print("the difference between A and B is", A-B)

# "^" will output all values contained in either A or B, but not both
print("the symmetric difference of A and B is", A^B)

# "tuple" will recast A to be a tuple. Here, A is converted from a set to a tuple
# sets have brackets: {} while tuples have brackets: ()
print("A displayed as a tuple is", tuple(A))

```

## Problem 3 — Simple phonebook (dict)
Create a small dictionary mapping names mapping food items to a personal score, say out of  100 (3–5 entries). Then:
- Add a new entry
- Update an existing entry
- Look up a score safely (avoid KeyError) (print the output of looking up an key not in the dictionary).
- Remove one entry

Print your dictionary after each operation and print the output

```python
# TODO: Create and manipulate a dictionary per the instructions.

#creating a list where strings (food items) are assigned numeric values
foodcritic = { "croissant": 90, "sushi": 95, "milk" :30, "cookies":60,}
print("the food critic rates the following foods", foodcritic)

#Adding a new food by assigning a string a variable within the pre-existing list
foodcritic["centro pasta"]= -100
print("food critic tries centro pasta, the rating is:", foodcritic.get("centro pasta"))

#updating an existing food rating with .update
foodcritic.update({"milk": 45,})
print("the food ratings have been updated. The new ratings are:", foodcritic)

#looking up a rating using .get
print("the rating for sushi is:", foodcritic.get("sushi"))

#looking up a food not in ratings
#prints none as it is not contained in list
print("the rating for pizza is:", foodcritic.get("pizza"))

# removing an item from list using.pop
print("croissant has been removed from the rating list", foodcritic.pop("croissant"))
print("the new rating list is:", foodcritic)

```
