ROOT FOR EQUATION  1

```python
#importing numpy library for equation computation
import numpy as np

#setting the intervals
a=float(input("enter a value for the lower limit of the interval [a,b]:"))
b=float(input("enter a value for the upper limit of the interval [a,b]:"))
print("the interval is [",a,",",b,"]")

# max_iter sets how many times a loop will run
max_iter = 1000

#tolerance sets the "error margin" for a calculated value
tolerance = 1e-10

#defining the functions
#the input in my_function  will select which function f(x) will be defined by
my_function = input (" choose a function to find the root of: \n 1. f(x)= e^x/10 + ln(x/10) \n 2. f(x)= arctan(x) - x^2 \n 3. f(x)= sin(x)/ln(x) \n 4. f(x)= ln(cos(x)) \n")
def f(x):
    #setting return values for function f(x) based on inputs within my_function
    if my_function == "1":
        return np.exp(x)/10 + np.log(x/10)
    elif my_function == "2":
        return np.arctan(x) - x**2
    elif my_function == "3":
        return np.sin(x)/np.log(x)
    elif my_function == "4":
        return np.log(np.cos(x))
    else:
        print("invalid function choice!!")

#printing the chosen function
if my_function == "1":
    print("the chosen function was f(x)= e^x/10 + ln(x/10)")
elif my_function == "2":
    print("the chosen function was f(x)= arctan(x) - x^2")
elif my_function == "3":
    print("the chosen function was f(x)=sin(x)/ln(x)")
elif my_function == "4":
    print("the chosen function was f(x)=ln(cos(x))")


#boolean identifying wether a root has been found
root_approximated = False

#this will show the approximate root if it is found
approximate_root = None

#failure reason: 
bisection_failure = False
bisection_failure_reason = ""

# this value tracks how many iterations of a loop have run
i = 0

#assigning left and right endpoints based on values inputted earlier
L=a
R=b

# a and b must be opposite signs, else the interval does not cross the x-axis and the bisection method cannot apply
if f(L) * f(R) > 0:
    bisection_failure = True
    bisection_failure_reason = "bisection method does not apply: f(a) and f(b) have the same sign"

#if the value of the function at the upper limit of the interval is equal to zero, the upper limit must be a root
elif f(R)==0:
    root_approximated = True
    approximate_root = R

#if the value of the function at the lower limit of the interval is equal to zero, the lower limit must be a root
elif f(L) ==0:
    root_approximated = True
    approximate_root = L

#creating a loop to find the roots in any other situation
else:
    #loop will run if the root_approximated boolean remains false and there has been less than 1000 loop repetitions
    while not root_approximated and i < max_iter:
       
        #assigning a variable to midpoint equation
        midpt = (L+R)/2

        #if the value of the function at the midpoint is 0 or the difference in the interval values is less than the tolerance
        # y=0 and the root has been found
        if f(midpt) == 0 or abs(R - L) < tolerance:
            root_approximated = True
            approximate_root = midpt

        #if root is not found, interval must be narrowed
        #if L and the midpoint are the same sign, the sign difference must lie in R
        # set L as the new midpoint, narrowing the interval in the direction of R
        elif f(L)*f(midpt)>0:
           L=midpt

        #otherwise L and the mipoint are differing signs
        # set R as the new midpoint to narrow interval and keep sign change
        else:
           R=midpt
        

        #increases counter for iterations by one for each loop
        i += 1

#printing final results   
print ("the root of your function is", approximate_root)
if bisection_failure == True:
    print(bisection_failure_reason)
print ("the number of iterations it took to find the root was", i)
```
