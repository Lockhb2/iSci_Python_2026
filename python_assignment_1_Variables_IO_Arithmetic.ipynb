## Problem 2 — Simple calculator
Ask the user for two numbers and print:
- Sum, difference, product, quotient
- Floor division and remainder
- One power example (e.g., a**b)

```python

# TODO: Implement a small calculator using input(). Be sure to cast input to numbers (int or float).
# Hints:
# a = float(input("Enter a number: "))
# b = float(input("Enter another number: "))

#making my calculator text
print("This is a small calculator. It can perform addition, subtraction, multiplication, and division.")

#allowing input of 2 numbers (decimals allowed)
a = float(input("Enter a number:"))

b = float(input("Enter another number:"))

print("you have selected numbers", a, "and", b)

# addition
print("the sum of", a, "and", b, "is:", a+b)

#subtraction
print("the difference of", a, "and", b,"is:", a-b)

#multiplication
print("the product of", a, "and", b, "is:", a*b)

#division (no remainders)
print("the quotient of", a, "and", b, "is:", a/b)

#division (integers)
print("the integer quotient of ", a, "and", b, "is:", a//b)

#division (remainders)
print("the remainder of", a, "and", b, "is:", a%b)

#powers
print(a, "raised to the power of", b, "is:", a**b)



```
