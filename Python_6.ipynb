## Problem 1 — Slicing practice
For `s = "calculus with python"`, print:
- First 8 characters
- Last 6 characters
- Every second character
- Reverse the string

```python

# TODO: Perform and print the described slices on s.
s = "calculus with python"
# Your code here...

#assigning s to a string value
s=str(input("enter a string"))
print("your string is:", s)

#printing a slice of the first 8 characters
# start at 0, not 1
print("the first 8 characters are:",s[0:8])

#print a slice of the last 6 characters
# negative sign (used for negative indexing) starts at the end of the string
print("the last 6 characters are:", s[-6:]) 

# printing every second character in the string
#:: is used to print every nth character, here ::2 is inputted, so every 2nd character is printed
print("every second character in the string is:", s[::2])

# reversing the string using
# ::-1 will print every 1st number(so every number) starting from the end of the string
print("the reverse of the string is:", s[::-1])



```

## Problem 2 — 2D list indexing
Given a 3x3 list `M` below, (1) print the diagonal elements only, then (2) replace the middle element with 0 and (3) print all of `M`.

```python

# TODO: Index into rows/columns and modify the middle element.
M = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]
# Your code here...
#printing original matrix63
.
print("the original matrix is :", M)

#printing diagonal elements of matrix
#use M[row][column] to print a specific row and column value (recall,indexing starts at 0!!)
print("the diagonal elements are :", M[0][0], M[1][1], M[2][2])

#replacing middle element with 0
#middle element is row 1 column 1, so M[1][1] is replaced with one
M[1][1] = 0
print("the middle number has been replaced with 0")
print(" the new matrix is:", M)
```

## Problem 3 — List comprehensions

- The list containing the first 100 odd numbers ($[1,3,5, ...]$).
- From `words = ["apple", "a", "banana", "pi", "grape"]`, create a new list of words, where all the words have length $\ge 3$

```python

# TODO: Implement the comprehensions.
# print numbers (X) in the set range (0-100) that result in a remainder of 1 when divided bty 2 (i.e odd numbers)
odds = [x for x in range(0, 100) if x % 2 == 1]

# defining a list of words (added "pen" to check if >=3 works)
words = ["apple", "a", "banana", "pi", "grape", "pen"]
print(" the set of words are:", words)

# crating a new list of words with length greater than or equal to 3
long_words = [words for words in words if len(words) >= 3]

#len will return the number of variables, aka "length" of a list
print("the number of odd numbers from 1-> 100 is:", (len(odds)))
print("the odd numbers are:", (odds))
print(long_words)
```
