## Problem 1 — Sum of first N square integers
Ask the user for `N` and compute the sum $1^2 + 2^2 + ... + N^2$ using a loop.

```python
#defining n
n=int(input("enter a value for n:"))
print("the value of n is:",n)
print("this code will calculate the sum of the squares of the integers from 1 to", n)

#defining a function to calculate the sum of n integers
def sum_of_n_integers(n):
    return sum(i**2 for i in range(1,n+1))

# printing final value
print("the sum of the squares of the first", n, "integers is", sum_of_n_integers(n),)

    
    

```

## Problem 3 — Guessing game (while loop)

- If the guess is too low/high, print a hint.
- Count the number of guesses.
